```python
def plot_tas_ensemble_2000(ds, n_cols=3, figsize_per_plot=(5, 4)):
    """
    Plot tas in year 2000 across space for all ensemble members.
    
    Parameters:
    -----------
    ds : xarray.Dataset
        Dataset containing 'tas' variable with dimensions including 'member_id', 'lat', 'lon', 'year'
    n_cols : int
        Number of columns in subplot grid (default: 3)
    figsize_per_plot : tuple
        Figure size for each subplot in inches (default: (5, 4))
    """
    # Select data for year 2000
    data_2000 = ds['tas'].sel(year=2000)
    
    # Get unique ensemble members
    members = data_2000.member_id.values
    n_members = len(members)
    
    # Calculate grid dimensions
    n_rows = int(np.ceil(n_members / n_cols))
    figsize = (n_cols * figsize_per_plot[0], n_rows * figsize_per_plot[1])
    
    # Create figure with subplots
    fig, axes = plt.subplots(n_rows, n_cols, figsize=figsize, 
                              subplot_kw=dict(projection=ccrs.PlateCarree()))
    axes = axes.flatten()  # Flatten to 1D for easier iteration
    
    # Plot each ensemble member
    for idx, member in enumerate(members):
        ax = axes[idx]
        
        # Select data for this member
        member_data = data_2000.sel(member_id=member)
        
        # Plot the data
        member_data.plot(ax=ax, transform=ccrs.PlateCarree(), cbar_kwargs={'shrink': 0.8})
        
        # Add geographic features
        ax.add_feature(cfeature.COASTLINE, linewidth=0.5)
        ax.add_feature(cfeature.BORDERS, linewidth=0.3)
        ax.add_feature(cfeature.LAND, facecolor="lightgray", alpha=0.3)
        ax.set_title(f'Member: {member}', fontsize=10)
    
    # Hide unused subplots
    for idx in range(n_members, len(axes)):
        axes[idx].set_visible(False)
    
    plt.suptitle('Temperature in 2000 - All Ensemble Members', fontsize=14, y=0.995)
    plt.tight_layout()
    
    if save_figs:
        plt.savefig(make_figure_filename('tas_ensemble_2000', 'png', 'figs/'), dpi=150, bbox_inches='tight')
    plt.show()

# Call the function
plot_tas_ensemble_2000(ds_max_masked)
```

```python
def check_lat_lon_grids_consistent():
    """
    Check if all CMIP6 tas_annual_max landonly datasets have consistent lat/lon grids.

    Returns
    -------
    bool
        True if all grids are identical, False otherwise.
    """
    data_folder = DATA_ROOT / 'CMIP6' / 'tas_annual_max'
    fnames = list(data_folder.glob("*.nc"))
    
    if not fnames:
        print("No *.nc files found in the data folder.")
        return False
    
    # Open the first dataset to get reference lat/lon
    ds_first = xr.open_dataset(fnames[0])
    lat_ref = ds_first.lat
    lon_ref = ds_first.lon
    ds_first.close()  # Close to save memory
    
    consistent = True
    for fname in fnames[1:]:
        ds = xr.open_dataset(fname)
        if not ds.lat.equals(lat_ref) or not ds.lon.equals(lon_ref):
            print(f"Grid mismatch in {fname.name}")
            consistent = False
        ds.close()
    
    if consistent:
        print("All datasets have consistent lat/lon grids.")
    else:
        print("Some datasets have mismatched lat/lon grids.")
    
    return consistent
```

## Algorithm: Figure 1
1. Take the hottest year on record (already done in preprocessing)
2. Fit a GEV to each (lat, lon) pair
3. Compute the Kuiper statistic at each (lat, lon)
4. Aggregate and plot CDF of Kuiper statistic (from obs)
5. Take 75 samples from the fitted GEV distributions
6. Aggregate synthetic data and plot CDF of Kuiper statistic (From synthetics)
7. Plot against reanalysis
8. Hopefully, it'll all be the same.
9. Repeat for each grid size and max/min data.

```python
def get_gev_fit_lat_lon(da, fit_dim, parallel=True):
    
    def _gev_helper(values):
        values = values[np.isfinite(values)]
        if len(values) < 10:
            return np.array([np.nan] * 3)
        try:
            c, loc, scale = genextreme.fit(values)
            return np.array([c, loc, scale])
        except Exception:
            return np.array([np.nan] * 3)

    if parallel:
        gev_params = xr.apply_ufunc(
            _gev_helper,
            da,
            input_core_dims=[['year']],
            output_core_dims=[['gev_params']],
            vectorize=True,
            dask='parallelized',
            output_dtypes=[float],
        )
    else:
        gev_params = xr.apply_ufunc(
            _gev_helper,
            da,
            input_core_dims=[['year']],
            output_core_dims=[['gev_params']]
        )

    #gev_params = gev_params.assign_coords(gev_param=['shape', 'location', 'scale'])
    #gev_ds = gev_params.to_dataset(dim='gev_params')

    return gev_params
```

```python
def _synth_sampler(shape, loc, scale):
    if np.isnan(shape) or np.isnan(loc) or np.isnan(scale):
        return np.full(75, np.nan)
    return genextreme.rvs(shape, loc, scale, size=75)

ds_synth = xr.apply_ufunc(
        _synth_sampler,
        ds_max_masked['shape'],
        ds_max_masked['loc'],
        ds_max_masked['scale'],
        input_core_dims=[[], [], []],
        output_core_dims=[['samples']],
        vectorize=True,
        dask='parallelize',
        output_dtypes=[float]
    )
```

```python
def _kuiper(sample, shape, loc, scale):
    sample = sample[np.isfinite(sample)]

    if len(sample) < 10:
        return np.array([-1])

    else:
        k, _ = kuiper(sample,
                           lambda x: genextreme.cdf(x,
                                                c=shape, loc=loc, scale=scale))
        return np.array([k]) 

da_k = xr.apply_ufunc(
    _kuiper,
    ds_max_masked['t2m'],
    ds_max_masked['shape'],
    ds_max_masked['loc'],
    ds_max_masked['scale'],
    input_core_dims=[['year'], [], [], []],
    output_core_dims=[['kuiper']],
    vectorize=True,
    dask='parallelize',
    output_dtypes=[float]
)

da_ks = xr.apply_ufunc(
    _kuiper,
    ds_max_masked['synth_t2m'],
    ds_max_masked['shape'],
    ds_max_masked['loc'],
    ds_max_masked['scale'],
    input_core_dims=[['year'], [], [], []],
    output_core_dims=[['kuiper']],
    vectorize=True,
    dask='parallelize',
    output_dtypes=[float]
)
```
