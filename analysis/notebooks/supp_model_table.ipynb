# Table of climate models and members that we use

```python
# import base packages for data analysis
import pandas as pd

from evt_heat_waves.config import DATA_ROOT, CONFIG_PATH
from evt_heat_waves.mip_fit.cmip_dataclass import CMIP6EnsembleConfig
```

```python
CMIPConfig = CMIP6EnsembleConfig.from_yaml(
    CONFIG_PATH.parent / 'meta.yaml',
    CONFIG_PATH.parent /  'qc.yaml'
)

AMIPConfig = CMIP6EnsembleConfig.from_yaml(
    CONFIG_PATH.parent / 'meta_amip.yaml',
    CONFIG_PATH.parent / 'qc_amip.yaml'
)
```

```python
rows = []
rows_amip = []

for m in CMIPConfig.iter_active_models('tas_annual_max'):
    tmp_prim_mem = m.primary_member
    prim_mem = tmp_prim_mem.split("_")[1]

    rows.append({
        "Model": m.name,
        "Variables": "tas_annual_max, tas_annual_mean",
        "Primary Member": prim_mem,
        "Time Interval": "1979-2024",
        "Experiment": "historical + SSP5-8.5"
    })

for m in AMIPConfig.iter_active_models('tas_annual_max'):
    tmp_prim_mem = m.primary_member
    prim_mem = tmp_prim_mem.split("_")[1]

    rows_amip.append({
        "Model": m.name,
        "Variables": "tas_annual_max, tas_annual_mean",
        "Primary Member": prim_mem,
        "Time Interval": "1979-2024",
        "Experiment": "historical + SSP5-8.5"
    })
```

```python
df = pd.DataFrame(rows)
df_amip = pd.DataFrame(rows_amip)
df_amip.sort_values("Model")
```

```python
df.sort_values("Model")
```

```python
pd.merge(df, df_amip, on='Model', how='outer').fillna('N/A').sort_values("Model")
```

```python
df.to_csv(
    DATA_ROOT / 'CMIP6' / 'supplemental_info' / 'model_var_primmem_time_exp.csv',
    index=False
)

df_amip.to_csv(
    DATA_ROOT / 'AMIP' / 'supplemental_info' / 'model_var_primmem_time_exp.csv',
    index=False
)
```
